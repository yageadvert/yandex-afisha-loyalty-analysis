```python
# Для выгрузки используем библиотеку SQLAlchemy.
import pandas as pd
import psycopg2
from sqlalchemy import create_engine

# Загружаем библиотеки для визуализации данных
import matplotlib.pyplot as plt
import seaborn as sns

# Загружаем библиотеку для расчёта коэффициента корреляции phi_k
from phik import phik_matrix

# Импорт библиотеки для работы с окружением
import os
# Импорт load_dotenv
from dotenv import load_dotenv
```

```python
# Загружаем переменные из файла .env
load_dotenv()

# Подключаемся к базе данных `data-analyst-afisha`:
db_config = {'user': os.getenv('DB_USER'), # имя пользователя
             'pwd': os.getenv('DB_PASSWORD'), # пароль
             'host': os.getenv('DB_HOST'),
             'port': os.getenv('DB_PORT'), # порт подключения
             'db': os.getenv('DB_NAME'), # название базы данных
            }

connection_string = 'postgresql://{}:{}@{}:{}/{}'.format(
    db_config['user'],
    db_config['pwd'],
    db_config['host'],
    db_config['port'],
    db_config['db'],
)

engine = create_engine(connection_string)
```

```python
# SQL-запрос, выгружающий в датафрейм pandas необходимые данные:
query = '''
-- Настройка параметра synchronize_seqscans важна для проверки
WITH set_config_precode AS (
    SELECT set_config('synchronize_seqscans', 'off', true)
)
SELECT
    user_id,
    device_type_canonical,
    order_id,
    created_dt_msk AS order_dt,
    created_ts_msk AS order_ts,
    currency_code,
    revenue,
    tickets_count,
    (p.created_dt_msk::date - LAG(p.created_dt_msk::date) OVER (
        PARTITION BY p.user_id
        ORDER BY p.created_dt_msk
    ))::int AS days_since_prev,
    event_id,
    event_type_main,
    service_name,
    region_name,
    city_name
FROM afisha.purchases AS p
JOIN afisha.events e USING (event_id)
JOIN afisha.city c USING (city_id)
JOIN afisha.regions r USING (region_id)
CROSS JOIN set_config_precode
WHERE p.device_type_canonical IN ('mobile', 'desktop') AND e.event_type_main NOT IN ('фильм')
ORDER BY p.user_id
'''
```

```python
df = pd.read_sql_query(query, con=engine)
```

Данные загружены и сохранены в датасет `df` без ошибок

Для этого используем датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Загрузим его по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv`

**Загружаем новый датасет к курсом валют:**

```python
df_tenge=pd.read_csv('https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')
```

```python
# Меняем тип данных столбц `data`, по которому будем присоединять таблицу
df_tenge['data']=pd.to_datetime(df_tenge['data'])

# Проверяем тип данных
df_tenge['data'].info()
```

**Соединяем датасеты для расчетов:**

```python
# Присоединяем два столбца из таблицы с курсом - `data`, `curs`
df=df.merge(df_tenge[['data', 'curs', 'nominal']], left_on='order_dt', right_on='data', how='left')
```

```python
# Сперва проверим уникальность данных в столбце с обозначением валюты
df['currency_code'].unique()
```

```python
# Добавляем новый столбец `revenue_rub` с условием: если сумма сделки была в kzt, переводим в rub
for i, val in enumerate(df['currency_code']):    
    if val == 'kzt':
        df.loc[i, 'revenue_rub'] = (df.loc[i, 'revenue'] * df.loc[i, 'curs'] / df.loc[i, 'nominal'])
    else:
        df.loc[i, 'revenue_rub'] = df.loc[i, 'revenue']

# Округлим значения в новом столбце до сотых долей:
df['revenue_rub']=df['revenue_rub'].round(2)
```

```python
# Проверим получившиеся значения
df[df['currency_code'] == 'kzt'].head()
```

```python
# Удалим столбцы, которые использовались в расчетах и не понадобятся в анализе
df = df.drop(['currency_code', 'revenue', 'data', 'curs', 'nominal'], axis=1)

# Получившаяся таблица
df.head()
```

### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас построим агрегированные признаки, описывающие поведение и профиль каждого пользователя.

---

**Задача 3.1.** Построим профиль пользователя — для каждого пользователя найдем:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавим два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.


---

**Агрегируем данные по пользователям:**

```python
# Сортируем данные по пользователям и времени совершения заказа
df_users=df.sort_values(['user_id','order_ts'])
```

```python
# Проводим группировку по пользователям
df_users = df_users.groupby('user_id').agg(
    first_order_dt = ('order_dt', 'min'), # дата первого заказа
    last_order_dt = ('order_dt', 'max'), # дата последнего заказа
    first_device_type = ('device_type_canonical', 'first'), # устройство, с которого был сделан первый заказ
    first_region = ('region_name', 'first'), # регион, в котором был сделан первый заказ
    first_service = ('service_name', 'first'), # билетный партнёр, к которому обращались при первом заказе
    first_event_type = ('event_type_main', 'first'), # жанр первого посещённого мероприятия
    total_orders = ('order_id', 'count'), # общее количество заказов
    avg_revenue = ('revenue_rub', 'mean'), # средняя выручка с одного заказа в рублях
    avg_tickets_count = ('tickets_count', 'mean'), # среднее количество билетов в заказе
    avg_days_since = ('days_since_prev', 'mean') # среднее время между заказами
)
```

```python
# Сбрасываем индекс со столбца user_id
df_users = df_users.reset_index()

# Выводим результат
df_users
```

```python
# Округлим средние значения до двух знаков после запятой
df_users[['avg_revenue', 'avg_tickets_count', 'avg_days_since']] = df_users[['avg_revenue', 'avg_tickets_count', 'avg_days_since']].round(2)
```

```python
# Добавляем новые столбцы 
df_users['is_two'] = (df_users['total_orders'] >= 2)
df_users['is_five'] = (df_users['total_orders'] >= 5)
```

```python
# Проверяем получившийся датасет
df_users
```

Построены профили пользователей и добавлены два бинарных признака.

#### 4.4. Корреляционный анализ количества покупок и признаков пользователя

Изучим какие характеристики первого заказа и профиля пользователя могут быть связаны с числом покупок. Для этого используем универсальный коэффициент корреляции `phi_k`, который позволяет анализировать как числовые, так и категориальные признаки.

---

**Задача 4.4.1:** Проведем корреляционный анализ:
- Рассчитаем коэффициент корреляции `phi_k` между признаками профиля пользователя и числом заказов (`total_orders`). При необходимости используем параметр `interval_cols` для определения интервальных данных.
- Проанализируем полученные результаты.
- Визуализируем результат корреляции с помощью тепловой карты.
- Ответим на вопрос: какие признаки наиболее связаны с количеством заказов?

---

```python
# Задаем все столбцы 'df_users', которые будут участвовать в корреляионном анализа
cols = [
    'total_orders',
    'first_device_type',
    'first_region',    
    'first_service',
    'first_event_type',
    'avg_revenue',
    'avg_tickets_count',
    'avg_days_since',
    'first_dayweek'
]
# Передаем столбцы в параметр `interval_cols` для определения интервальных данных
interval_cols = [
    'total_orders',
    'avg_revenue',
    'avg_tickets_count',
    'avg_days_since',
]

correlation_matrix=df_users[cols].phik_matrix(interval_cols=interval_cols)
correlation_matrix
```

```python
# Выводим из корреляционной матрицы один столбец 'order_count' для наглядности
correlation_orders_count=correlation_matrix.loc[correlation_matrix.index != 'total_orders'][['total_orders']].sort_values(by='total_orders',ascending=False)

# Строим хитмап по одному корреляционному столбцу
plt.figure(figsize=(3, 6));
sns.heatmap(correlation_orders_count,
            annot=True, # Отображаем численные значения в ячейках карты
            fmt='.2f', # Форматируем значения корреляции: два знака после точки
            cmap='coolwarm', # Устанавливаем цветовую гамму от красного (макс. значение) к синему
            linewidths=0.5 # Форматируем линию между ячейками карты
           );

# Добавляем заголовок и подпись по оси Х
plt.title('Корреляция количества заказов от других признаков')
plt.xlabel('');

# Выводим график
plt.show();
```

На данный момент в датасете df_users столбец `orders_segment` содержит пропуски в тех строках, где у клиента менее 2 заказов. Уберем пропуски, добавив сегмент "От 1 до 2" и используем его для корреляционного анализа:

```python
# Корректируем столбец orders_segment
df_users=df_users.copy()
df_users['orders_segment'] = pd.cut(df_users['total_orders'], bins=[1, 2, 5, float('inf')],
                                  labels=['От 1 до 2', 'От 2 до 4', 'От 5 и выше'], right=False)
df_users.head()
```

```python
# Задаем все столбцы 'df_users', которые будут участвовать в корреляионном анализа
cols = [
    'orders_segment',
    'first_device_type',
    'first_region',    
    'first_service',
    'first_event_type',
    'avg_revenue',
    'avg_tickets_count',
    'avg_days_since',
    'first_dayweek'
]
# Передаем столбцы в параметр `interval_cols` для определения интервальных данных
interval_cols = [
    'avg_revenue',
    'avg_tickets_count',
    'avg_days_since',
]

correlation_matrix_1=df_users[cols].phik_matrix(interval_cols=interval_cols)
correlation_matrix_1
```

```python
# Выводим из корреляционной матрицы один столбец 'orders_segment' для наглядности
correlation_orders_segment=correlation_matrix_1.loc[correlation_matrix_1.index != 'orders_segment'][['orders_segment']].sort_values(by='orders_segment',ascending=False)

# Строим хитмап по одному корреляционному столбцу
plt.figure(figsize=(3, 6));
sns.heatmap(correlation_orders_segment,
            annot=True, # Отображаем численные значения в ячейках карты
            fmt='.2f', # Форматируем значения корреляции: два знака после точки
            cmap='coolwarm', # Устанавливаем цветовую гамму от красного (макс. значение) к синему
            linewidths=0.5 # Форматируем линию между ячейками карты
           );

# Добавляем заголовок и подпись по оси Х
plt.title('Корреляция количества заказов, сгруппированных по сегментам, от других признаков')
plt.xlabel('');

# Выводим график
plt.show();
```

Наибольшую прямую корреляцию к количеству заказов пользователя имеют два параметра: среднее число билетов в заказе и среднее время между заказами.
