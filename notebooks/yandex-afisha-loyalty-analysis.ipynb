**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведем выручку к единой валюте — российскому рублю.

Для этого используем датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Загрузим его по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv`

Значения в рублях представлено для 100 тенге.

---


Перед началом присоединения таблиц, меняем в датасете `df_tenge` тип данных столбца `data` на datetime, который будет ключом:

```python
df_tenge['data']=pd.to_datetime(df_tenge['data'])
df_tenge['data'].info()
```

Датасет готов к присоединению к `df`

**Соединяем датасеты для расчетов:**

```python
# Присоединяем два столбца из таблицы с курсом - `data`, `curs`
df=df.merge(
    df_tenge[['data', 'curs', 'nominal']], 
    left_on='order_dt', 
    right_on='data', 
    how='left')

# Пересчитываем курсы валют в rub с помощью метода where и присоединенной таблицы:
df['revenue_rub'] = (df['revenue']
    .where(df['currency_code']=='rub', 
    df['revenue']*df['curs']/100)
    ).round(2)

# Проверим получившиеся значения:
df[df['currency_code'] == 'kzt'].head()
```

```python
# Удалим столбцы, которые использовались в расчетах и не понадобятся в анализе
df = df.drop(['currency_code', 'revenue', 'data', 'curs', 'nominal'], axis=1)

df.head()
```

Есть вероятность, что среди таких строк окажутся реальные заказы, которые пользователь совершил в тот же день и полностью идентичные его предыдущему заказу. Но в рамках поставленной задачи - исследования лояльности и возврата клиентов на площадку, эти случайности не исказят результатов анализа.

Из примера выше видно, что при удалении дубликатов, колонка `days_since_prev` станет не корректной, потому что дубликаты могут выпасть на пересечение следующих дней. После удаления дубликатов необходимо пересчитать этот столбец.

```python
(
    df
    .sort_values('order_ts')
    .drop_duplicates(subset='user_id', keep='first')
    .head(10)
)
```

Посчитаем количество уникальных пользователей в датасете и сравним их число с количеством пропусков в отсортированном списке пользователей:

```python
df['user_id'].nunique() == (
    df
    .sort_values('order_ts')
    .drop_duplicates(subset='user_id', keep='first')
    ['days_since_prev']
    .isna()
    .sum()
)
```

Значения NaN в столбце days_since_prev соответствуют ожидаемым - это первые заказы. 


Удаляем в датасете дубликаты:

```python
cols = [
    'user_id',
    'tickets_count',
    'event_id',
    'service_name',
    'revenue_rub'
]
df = df.drop_duplicates(subset=cols, keep="first")
```

После удаления дубликатов, колонка `days_since_prev` не всегда отражает корректное количество дней, поэтому пересчитываем столбец:

```python
df = df.copy()
df['days_since_prev'] = (
    df['order_dt'] - df
    .groupby('user_id')['order_dt']
    .shift(1)
).dt.days
```

### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас построим агрегированные признаки, описывающие поведение и профиль каждого пользователя.

---

**Задача 3.1.** Построим профиль пользователя — для каждого пользователя найдем:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавим два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.


---

```python
profiles = (df
        # В начале сортируем данные по дате совершения заказа, что найти первые признаки:
        .sort_values(by='order_ts')
        # Затем группируем по номеру пользователя и агрегируем данные:
        .groupby('user_id')
        .agg(
            # Находим первую и последнюю даты заказа:
            first_order_dt=('order_dt','min'),
            last_order_dt=('order_dt','max'),
            # Находим устройства, регион, название билетного партнера и жанр мероприятия первого заказа:
            first_device=('device_type_canonical','first'),
            first_region_name=('region_name','first'),
            first_service_name=('service_name','first'),
            first_event_type=('event_type_main','first'),
            # Подсчитваем количество заказов:
            total_orders=('order_id','nunique'),
            # Считаем статистику по заказам: средняя стоимость заказа, среднее количество билетов:
            avg_revenue_rub=('revenue_rub','mean'),
            avg_tickets_count=('tickets_count','mean'),
            # Считаем среднее количество дней между покупками:
            avg_days_since_prev=('days_since_prev','mean')
        )
        # Создаем два признака: совершил ли пользователь 2 / 5 и более заказов:
        .assign(
            avg_revenue_rub = lambda x: x['avg_revenue_rub'].round(2),
            avg_tickets_count = lambda x: x['avg_tickets_count'].round(2),
            avg_days_since_prev = lambda x: x['avg_days_since_prev'].round(2),
            is_two = lambda x: x['total_orders'] >= 2,
            is_five = lambda x: x['total_orders'] >= 5
        )
        .reset_index()
)
```

    - по региону проведения мероприятия из первого заказа;
    - по билетному оператору, продавшему билеты на первый заказ.
- Подсчитаем общее количество пользователей в каждом сегменте и их долю в разрезе каждого признака. Сегмент — это группа пользователей, объединённых определённым признаком, то есть объединённые принадлежностью к категории. Например, все клиенты, сделавшие первый заказ с мобильного телефона, — это сегмент.
- Ответим на вопрос: равномерно ли распределены пользователи по сегментам или есть выраженные «точки входа» — сегменты с наибольшим числом пользователей?

---


Напишем пользовательскую функцию для дальнейшей группировки пользователей:

```python
def group_user(df,column):
    result = (
        df
        .groupby(column, observed=True)
        .agg(
            count_users=('user_id','count')
        )
        .sort_values('count_users',ascending = False)
# Добавляем новый столбец с процентной долей каждой строки
        .assign(
            percent_users = lambda x: (
                x['count_users'] / x['count_users'].sum()*100
            ).round(2)
        )
    )
    display(result)
    return result
```

**Задача 4.1.2.** Проанализируем возвраты пользователей:

- Для каждого сегмента вычислим долю пользователей, совершивших два и более заказа.
- Визуализируем результат подходящим графиком. Если сегментов слишком много, то поместим на график только 10 сегментов с наибольшим количеством пользователей. Такое возможно с сегментами по региону и по билетному оператору.
- Ответим на вопросы:
    - Какие сегменты пользователей чаще возвращаются на Яндекс Афишу?
    - Наблюдаются ли успешные «точки входа» — такие сегменты, в которых пользователи чаще совершают повторный заказ, чем в среднем по выборке?

Сперва считаем общую возвращаемость по всей выборке, чтобы использовать это значение в качестве сравнительной аналитики на графиках. Сохраним значение в переменную `overall_is_two_percent`, ее значение понадобится нам для анализа по всему дальнейшему проекту.

```python
overall_is_two_percent = profiles['is_two'].mean() * 100
```

Напишем пользовательскую функцию для расчета доли пользователей, вернувшихся за покупками для каждого сегмента:

```python
def group_istwo(df, column):
    result = (
        df
        .groupby(column, observed=True)
        .agg(
            count_users=('user_id','count'),
            is_two_percent=('is_two', lambda x: (x.mean() * 100).round(2))
        )
    )
    display(result)
    return result
```

**Анализируем долю вернувшихся пользователей, сегментированных по их первому мероприятию `first_event_type`:**

```python
istwo_first_event = group_istwo(profiles, ['first_event_type'])
```

```python
plt.figure(figsize=(7, 7))
order_event_istwo=istwo_first_event.sort_values('is_two_percent', ascending=False).index
# Строим график
ax = sns.barplot(
    data=istwo_first_event,
    x=istwo_first_event.index,
    y='is_two_percent',
    order=order_event_istwo,
    edgecolor='black'
)

# Строим добавочную линию в виде общей возвращаемости по всей выборке
plt.axhline(
    y=overall_is_two_percent,
    color='red',
    linestyle='--',
    linewidth=1,
    label=f'Общая средняя возвращаемость: {overall_is_two_percent:.1f}%'
)
# Добавляем подписи с процентами
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%')

plt.grid(axis='y', alpha=0.3)
plt.xlabel('Тип первого мероприятия')
plt.ylabel('Возвращаемость, %')
plt.title('Возвращаемость по типу первого мероприятия')
plt.legend()
plt.show()
```

- Выше среднего возвращаются пользователи, начавшие с выставок — 61.0%, театра — 59.2%, стендапа — 57.1% и концертов — 57.0%.

**Задача 4.2.3.** Проанализируем влияние среднего количества билетов в заказе на вероятность повторной покупки.

- Изучим распределение пользователей по среднему количеству билетов в заказе (`avg_tickets_count`) и опишем основные наблюдения.
- Разделим пользователей на несколько сегментов по среднему количеству билетов в заказе:
    - от 1 до 2 билетов;
    - от 2 до 3 билетов;
    - от 3 до 5 билетов;
    - от 5 и более билетов.
- Для каждого сегмента подсчитаем общее число пользователей и долю пользователей, совершивших повторные заказы.
- Ответим на вопросы:
    - Как распределены пользователи по сегментам — равномерно или сконцентрировано?
    - Есть ли сегменты с аномально высокой или низкой долей повторных покупок?

---

```python
profiles = profiles.copy()
profiles = profiles.assign(
    tickets_segment = lambda x: pd.cut(
        x['avg_tickets_count'], 
        bins=[1, 2, 3, 5, float('inf')],
        labels=['От 1 до 2', 'От 2 до 3', 'От 3 до 5', 'От 5 и выше'], 
        right=False)
    )
profiles.head()
```

```python
result423 = (profiles
    .groupby('tickets_segment', observed=True)
    .agg(
        count_users=('user_id','count'),
        count_is_two=('is_two','sum')
    )
    .assign(
         share_is_two=lambda x: (x['count_is_two'] / x['count_users']*100).round(2)
    )
            )
result423
```

```python
plt.figure(figsize=(10,5))
mean_1=result423['share_is_two'].mean()

ax=sns.barplot(
    data=result423,
    x=result423.index,
    y='share_is_two',
    edgecolor='black'
)

plt.axhline(
    y=mean_1,
    color='red',
    linestyle='--',
    linewidth=1,
    label=f'Средняя доля возвратов по сегментам: {mean_1:.1f}%'
)

# Добавляем подписи с процентами
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=1)

plt.title('Влияние среднего количества билетов в заказе на вероятность повторной покупки')
plt.xlabel('Сегмент количества билетов')
plt.ylabel('Доля возвратов пользователей')
plt.grid(axis='y', alpha=0.3)
plt.legend()
plt.show()
```

**Промежуточные выводы:**
- Пользователи, заказывающие от 2 до 3 билетов возвращаются на площадку чаще других - в 69% случаев. На втором месте пользователи с заказами от 3 до 5 билетов.
- Реже всего (13.5%) возвращаются пользователи с 5 и более билетами. Можно предположить, какой тип пользователей стоит за этими сегментами и скорректировать маркетинговые кампании в зависимости от индивидуальных предпочтений.

**Задача 4.3.1.** Проанализируем, как день недели, в которой была совершена первая покупка, влияет на поведение пользователей.

- По данным даты первого заказа выделим день недели.
- Для каждого дня недели подсчитаем общее число пользователей и долю пользователей, совершивших повторные заказы. Результаты визуализируем.
- Ответим на вопрос: влияет ли день недели, в которую совершена первая покупка, на вероятность возврата клиента?

---


```python
# Создаем для анализа новую копию profiles под названием result431:
result431 = (profiles
    .copy()
    .assign(
    # Добавляем новый столбец first_dayweek с номером для недели:
    first_dayweek = lambda df: df['first_order_dt'].dt.day_name(),
    
    # Добавляем столбец для сортировки:
    day_num = lambda df: df['first_order_dt'].dt.dayofweek
    )
    
    # Группируем данные по номеру дня и его названию:
    .groupby(['day_num','first_dayweek'], as_index=False)
    .agg(
        count_users=('user_id','count'),
        count_is_two=('is_two','sum')
    )

    # Добавляем расчет доли:
    .assign(
    share = lambda df: df['count_is_two'] / df['count_users']*100
    )
    .round(2)
    
    .sort_values('day_num')
    
    # Удаляем столбец для сортировки:
    .drop(columns='day_num')
    
    # Возвращаем индекс
    .set_index('first_dayweek')
)
result431
```

#### 4.4. Корреляционный анализ количества покупок и признаков пользователя

Изучим какие характеристики первого заказа и профиля пользователя могут быть связаны с числом покупок. Для этого используем универсальный коэффициент корреляции `phi_k`, который позволяет анализировать как числовые, так и категориальные признаки.

---

**Задача 4.4.1:** Проведем корреляционный анализ:
- Рассчитаем коэффициент корреляции `phi_k` между признаками профиля пользователя и числом заказов (`total_orders`). При необходимости используем параметр `interval_cols` для определения интервальных данных.
- Проанализируем полученные результаты.
- Визуализируем результат корреляции с помощью тепловой карты.
- Ответим на вопрос: какие признаки наиболее связаны с количеством заказов?

---

На данный момент в датасете df_users столбец `orders_segment` содержит пропуски в тех строках, где у клиента менее 2 заказов. Уберем пропуски, добавив сегмент "От 1 до 2" и используем его для корреляционного анализа:

```python
# Корректируем столбец orders_segment
profiles=profiles.copy()
profiles['orders_segment'] = pd.cut(profiles['total_orders'], bins=[1, 2, 5, float('inf')],
                                  labels=['От 1 до 2', 'От 2 до 4', 'От 5 и выше'], right=False)
profiles.head()
```

```python
# Задаем все столбцы 'df_users', которые будут участвовать в корреляионном анализа
cols = [
    'orders_segment',
    'first_device',
    'first_region_name',    
    'first_service_name',
    'first_event_type',
    'avg_revenue_rub',
    'avg_tickets_count',
    'avg_days_since_prev'
]
# Передаем столбцы в параметр `interval_cols` для определения интервальных данных
interval_cols = [
    'avg_revenue_rub',
    'avg_tickets_count',
    'avg_days_since_prev',
]

correlation_matrix_1=profiles[cols].phik_matrix(interval_cols=interval_cols)
correlation_matrix_1
```

```python
# Выводим из корреляционной матрицы один столбец 'orders_segment' для наглядности
correlation_orders_segment = (
    correlation_matrix_1
    .loc[correlation_matrix_1.index != 'orders_segment'][['orders_segment']]
    .sort_values(by='orders_segment',ascending=False)
)

# Строим хитмап по одному корреляционному столбцу
plt.figure(figsize=(3, 6));

sns.heatmap(
    correlation_orders_segment,

    # Отображаем численные значения в ячейках карты
    annot=True, 
    
    # Форматируем значения корреляции: два знака после точки
    fmt='.2f',

    # Устанавливаем цветовую гамму от красного (макс. значение) к синему
    cmap='viridis',

    # Форматируем линию между ячейками карты
    linewidths=0.5 ,
    vmin=0, 
    vmax=1
)

plt.title('Корреляция количества заказов,\n сгруппированных по сегментам, от других признаков')
plt.xlabel('')

# Выводим график
plt.show()
```

- Средний интервал между заказами — 0.39. Связь логична: чем чаще пользователь совершает заказы, тем меньше в среднем промежуток между ними. Этот признак в большей степени отражает уже сформированную активность, а не только изначальные характеристики пользователя.

**Действие:** управление привлечением новых пользователей исходя из полученных статистических данных - не запускать массовые программы лояльности только по устройству, региону или дню недели, а анализировать полученные точки входа и тестировать новые точечные решения.

**Ожидаемый эффект:** приход релевантных пользователей с лучшим первым взаимодействие и контроль рынка привлечения новых клиентов, но без ожидания значимого роста повторных покупок исключительно за счёт этих признаков.

**Общий итог**

Необходимо строить отдельные маркетинговые программы по привлечению новых клиентов, исходя из устройств, региона, билетного оператора, типа события и деня недели, и отдельные программы для управления возвращаемостью пользователей, акцентируя внимание на 
количестве билетов, интервале между заказами и выручке, однако работать с ними нужно через проверяемые гипотезы и персонализированные по каждому пользователю параметры.
